**NETFLIX DATA ANALYSIS**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/netflix_titles_cleaned.csv', encoding='latin1')
display(df.head())
```

```python
print(df.shape)
print(df.columns)
print(df.info())
```

```python
df.isnull().sum()
```

```python
content_type = df['type'].value_counts()

content_type.plot(
kind='bar',
figsize=(6,4),
title='Movies vs TV Shows'
)

plt.show()
```

```python
df['rating'].value_counts().head(10).plot(
kind='bar',
figsize=(8,4),
title='Top Ratings'
)

plt.show()
```

```python
df['release_year'].value_counts().sort_index().plot(
kind='line',
figsize=(10,5),
title='Content Release Trend'
)

plt.show()
```

```python
df['country'].value_counts().head(10).plot(
kind='bar',
figsize=(8,4),
title='Top Countries'
)

plt.show()
```

```python
df['type'].value_counts().plot(
kind='pie',
autopct='%1.1f%%',
figsize=(6,6)
)

plt.ylabel('')
plt.show()
```

Key Insights from Python Analysis

📌 Movies dominate Netflix's content library.

📌 TV-MA is the most common content rating.

📌 Content production increased significantly after 2015.

📌 The United States contributes the largest number of titles.

📌 Python enabled efficient data exploration and visualization
